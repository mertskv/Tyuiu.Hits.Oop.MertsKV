<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Customer в C#, который будет представлять информацию о
клиентах или покупателях. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу Customer:
• Атрибуты: Идентификатор клиента (CustomerId), Имя (Name), Электронная
почта (Email).
• Методы:
o
o GetFullName(): метод для получения полного имени клиента.
o UpdateEmail(string newEmail): метод для обновления электронной
почты клиента.
o ViewProfile(): метод для просмотра профиля клиента.

Требования к производным классам:
1. VIPКлиент (VipCustomer): Должен содержать дополнительные атрибуты,
такие как Баланс лояльности (LoyaltyPoints). Метод ViewProfile() должен быть

переопределен для отображения дополнительной информации о VIP-
клиенте.

2. ОбычныйКлиент (RegularCustomer): Должен содержать дополнительные
атрибуты, такие как Дата регистрации (RegistrationDate).
Метод UpdateEmail() должен быть переопределен для добавления
информации о дате последнего обновления электронной почты.
3. ГрупповойКлиент (GroupCustomer) (если требуется третий класс): Должен
содержать дополнительные атрибуты, такие как Название группы
(GroupName). Метод GetFullName() должен быть переопределен для
отображения названия группы вместо имени клиента.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.

```
using System;
using System.Collections.Generic;


public delegate void CustomerUpdateHandler(string message);


public class Customer
{
    // поля
    protected int _CustomerId;
    protected string _Name;
    protected string _Email;
    protected string _PhoneNumber;   
    protected string _Address;       

    // событие
    public event CustomerUpdateHandler CustomerUpdate;

    // свойства
    public int CustomerId
    {
        get { return _CustomerId; }
        set { _CustomerId = value; }
    }

    public string Name
    {
        get { return _Name; }
        set { _Name = value; }
    }

    public string Email
    {
        get { return _Email; }
        set
        {
            if (!string.IsNullOrEmpty(value) && value.Contains("@"))
                _Email = value;
            else
                throw new ArgumentException("Некорректный email!");
        }
    }

    public string PhoneNumber
    {
        get { return _PhoneNumber; }
        set { _PhoneNumber = value; }
    }

    public string Address
    {
        get { return _Address; }
        set { _Address = value; }
    }

    
    public Customer(int id, string name, string email, string phone, string address)
    {
        CustomerId = id;
        Name = name;
        Email = email;
        PhoneNumber = phone;
        Address = address;
    }

    
    protected void NotifyUpdate(string message)
    {
        CustomerUpdate?.Invoke(message);
    }

    
    public virtual string GetFullName()
    {
        return Name;
    }

    public virtual void UpdateEmail(string newEmail)
    {
        Email = newEmail;
        NotifyUpdate($"Email клиента '{GetFullName()}' обновлён на {newEmail}");
    }

    public virtual void ViewProfile()
    {
        Console.WriteLine($"ID клиента: {CustomerId}, Имя: {GetFullName()}, Email: {Email}, Телефон: {PhoneNumber}, Адрес: {Address}");
    }

    
    public void UpdatePhoneNumber(string newPhoneNumber, string countryCode)
    {
        PhoneNumber = countryCode + newPhoneNumber;
        NotifyUpdate($"Телефон клиента '{GetFullName()}' обновлён на {PhoneNumber}");
    }

    public void UpdateAddress(string newAddress)
    {
        Address = newAddress;
        NotifyUpdate($"Адрес клиента '{GetFullName()}' обновлён на {newAddress}");
    }
}


public interface ILevelActions
{
    string GetLevelCustomer();
}


public class VipCustomer : Customer, ILevelActions
{
    private int _LoyaltyPoints;
    private string _MembershipLevel;

    public int LoyaltyPoints
    {
        get { return _LoyaltyPoints; }
        set { _LoyaltyPoints = value; }
    }

    public string MembershipLevel
    {
        get { return _MembershipLevel; }
        set { _MembershipLevel = value; }
    }

    public VipCustomer(int id, string name, string email, string phone, string address,
                       int loyaltyPoints, string membershipLevel)
        : base(id, name, email, phone, address)
    {
        LoyaltyPoints = loyaltyPoints;
        MembershipLevel = membershipLevel;
    }

    
    string ILevelActions.GetLevelCustomer()
    {
        return $"VIP клиент, уровень: {MembershipLevel}";
    }

    
    public void AddLoyaltyPoints(int points)
    {
        LoyaltyPoints += points;
        NotifyUpdate($"VIP клиенту '{GetFullName()}' начислено {points} баллов. Баланс: {LoyaltyPoints}");
    }

    
    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"   [VIP] Баланс лояльности: {LoyaltyPoints}, Уровень членства: {MembershipLevel}\n");
    }
}


public class RegularCustomer : Customer
{
    private string _RegistrationDate;
    private DateTime _LastEmailUpdate;

    public string RegistrationDate
    {
        get { return _RegistrationDate; }
        set { _RegistrationDate = value; }
    }

    public DateTime LastEmailUpdate
    {
        get { return _LastEmailUpdate; }
    }

    public RegularCustomer(int id, string name, string email, string phone, string countryCode,
                           string address, string registrationDate)
        : base(id, name, email, phone, address)
    {
        UpdatePhoneNumber(phone, countryCode);
        RegistrationDate = registrationDate;
    }

    
    public override void UpdateEmail(string newEmail)
    {
        base.UpdateEmail(newEmail);
        _LastEmailUpdate = DateTime.Now;
        Console.WriteLine($"Дата последнего обновления email: {_LastEmailUpdate}");
    }

    
    public void ShowRegistrationInfo()
    {
        Console.WriteLine($"Клиент '{GetFullName()}' зарегистрирован: {RegistrationDate}");
    }

    
    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"   [Обычный] Дата регистрации: {RegistrationDate}, Последнее обновление email: {_LastEmailUpdate}\n");
    }
}


public class GroupCustomer : Customer
{
    private string _GroupName;
    private List<Customer> customers = new List<Customer>();

    public string GroupName
    {
        get { return _GroupName; }
        set { _GroupName = value; }
    }

    public int MembersCount
    {
        get { return customers.Count; }
    }

    public GroupCustomer(int id, string groupName, string email, string phone, string address)
        : base(id, groupName, email, phone, address)
    {
        GroupName = groupName;
    }

    public void AddCustomer(Customer customer)
    {
        customers.Add(customer);
        NotifyUpdate($"Клиент '{customer.GetFullName()}' добавлен в группу '{GroupName}'");
    }

    
    public override string GetFullName()
    {
        return GroupName;
    }

    
    public void ShowMembers()
    {
        Console.WriteLine($"В группе '{GroupName}' состоит {MembersCount} клиентов:");
        foreach (var c in customers)
            Console.WriteLine($"   - {c.GetFullName()}");
    }

    
    public override void ViewProfile()
    {
        Console.WriteLine($"ID группы: {CustomerId}, Название группы: {GetFullName()}, Email: {Email}, Участников: {MembersCount}");
        foreach (var c in customers)
            c.ViewProfile();
    }
}


public class CustomerManager
{
    private List<Customer> customers = new List<Customer>();

    public void AddCustomer(Customer customer)
    {
        customers.Add(customer);
        Console.WriteLine($"Клиент '{customer.GetFullName()}' добавлен в менеджер.");
    }

    public void DisplayAllCustomers()
    {
        foreach (var c in customers)
            c.ViewProfile();
    }

    public IEnumerable<Customer> FilterByType<T>() where T : Customer
    {
        foreach (var c in customers)
            if (c is T)
                yield return c;
    }

    public IEnumerable<Customer> FilterByName(string part)
    {
        return customers.FindAll(c => c.GetFullName().ToLower().Contains(part.ToLower()));
    }
}


public class CustomerCollection<T> where T : Customer
{
    private List<T> _customers = new List<T>();

    public void Add(T customer)
    {
        _customers.Add(customer);
    }

    public void DisplayAll()
    {
        foreach (var c in _customers)
            c.ViewProfile();
    }

    public int Count
    {
        get { return _customers.Count; }
    }
}


VipCustomer vip = new VipCustomer(1, "Пётр Петров", "peter@example.com", "1234567890", "Улица 1", 150, "Gold");
RegularCustomer regular = new RegularCustomer(2, "Сергей Сергеев", "sergey@example.com", "0987654321", "+7", "Улица 2", "01.01.2024");
GroupCustomer group = new GroupCustomer(3, "Группа A", "group_a@example.com", "2233445566", "Улица 5");


vip.CustomerUpdate += message => Console.WriteLine($"[Уведомление VIP]: {message}");
regular.CustomerUpdate += message => Console.WriteLine($"[Уведомление Regular]: {message}");
group.CustomerUpdate += message => Console.WriteLine($"[Уведомление Group]: {message}");

Console.WriteLine("=== Действия с клиентами ===\n");


vip.UpdateEmail("new_peter@example.com");
regular.UpdateEmail("new_sergey@example.com");
Console.WriteLine();


vip.AddLoyaltyPoints(50);
Console.WriteLine();


group.AddCustomer(vip);
group.AddCustomer(regular);
Console.WriteLine();


CustomerManager manager = new CustomerManager();
manager.AddCustomer(vip);
manager.AddCustomer(regular);
manager.AddCustomer(group);
Console.WriteLine();


Console.WriteLine("=== Все клиенты ===\n");
manager.DisplayAllCustomers();


Console.WriteLine("=== Только VIP клиенты ===\n");
foreach (var c in manager.FilterByType<VipCustomer>())
    c.ViewProfile();

Console.WriteLine("=== Клиенты, чьё имя содержит 'Серг' ===\n");
foreach (var c in manager.FilterByName("Серг"))
    c.ViewProfile();


Console.WriteLine("=== Generic-коллекция ===\n");
CustomerCollection<Customer> collection = new CustomerCollection<Customer>();
collection.Add(vip);
collection.Add(regular);
collection.Add(group);
collection.DisplayAll();
Console.WriteLine($"Всего в generic-коллекции: {collection.Count}\n");


Console.WriteLine("=== Явная реализация ILevelActions ===\n");
ILevelActions levelAction = vip;
Console.WriteLine(levelAction.GetLevelCustomer());


Console.WriteLine();
regular.ShowRegistrationInfo();


Console.WriteLine();
group.ShowMembers();


static void InteractCustomers(Customer c1, Customer c2)
{
    Console.WriteLine($"Клиент {c1.GetFullName()} отправляет привет клиенту {c2.GetFullName()}.");
    Console.WriteLine($"Клиент {c2.GetFullName()} отвечает: Здравствуй, {c1.GetFullName()}!");
}

Console.WriteLine("\n=== Взаимодействие ===\n");
InteractCustomers(vip, regular);
```
