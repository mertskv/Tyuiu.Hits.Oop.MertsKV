```
using System;
using System.Collections.Generic;

public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return $"Компонент: {Name}, Цена: {Price} руб.";
    }

    public override string ToString()
    {
        return GetInfo();
    }

    public static bool operator ==(Component a, Component b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(Component a, Component b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is Component c) return this == c;
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}

public interface IInstallable
{
    void Install();
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public double FrequencyGHz { get; set; }

    public CPU(string name, decimal price, int cores, double frequency) : base(name, price)
    {
        Cores = cores;
        FrequencyGHz = frequency;
    }

    public override string GetInfo()
    {
        return $"CPU: {Name}, {Price} руб., {Cores} ядер, {FrequencyGHz} ГГц";
    }

    public void Install()
    {
        Console.WriteLine($"  {Name} установлен.");
    }
}

public class RAM : Component, IInstallable
{
    public int CapacityGB { get; set; }
    public string Type { get; set; }

    public RAM(string name, decimal price, int capacity, string type) : base(name, price)
    {
        CapacityGB = capacity;
        Type = type;
    }

    public override string GetInfo()
    {
        return $"RAM: {Name}, {Price} руб., {CapacityGB} ГБ {Type}";
    }

    public void Install()
    {
        Console.WriteLine($"  {Name} установлена.");
    }
}

public class GPU : Component, IInstallable
{
    public int VRAMGB { get; set; }
    public string Interface { get; set; }

    public GPU(string name, decimal price, int vram, string pciInterface) : base(name, price)
    {
        VRAMGB = vram;
        Interface = pciInterface;
    }

    public override string GetInfo()
    {
        return $"GPU: {Name}, {Price} руб., {VRAMGB} ГБ, {Interface}";
    }

    public void Install()
    {
        Console.WriteLine($"  {Name} установлена.");
    }
}

public class HDD : Component, IInstallable
{
    public int CapacityGB { get; set; }
    public string ConnectionType { get; set; }

    public HDD(string name, decimal price, int capacity, string connection) : base(name, price)
    {
        CapacityGB = capacity;
        ConnectionType = connection;
    }

    public override string GetInfo()
    {
        return $"HDD: {Name}, {Price} руб., {CapacityGB} ГБ, {ConnectionType}";
    }

    public void Install()
    {
        Console.WriteLine($"  {Name} подключен.");
    }
}

public class Computer
{
    private List<Component> _components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (_components.Remove(component))
            Console.WriteLine($"  {component.Name} извлечен.");
        else
            Console.WriteLine($"  {component.Name} не найден.");
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (var c in _components) total += c.Price;
        return total;
    }

    public void ShowAllComponents()
    {
        foreach (var c in _components)
            Console.WriteLine($"  {c.ToString()}");
    }
}

// Демонстрация
Console.WriteLine("Собираем ПК...\n");
CPU cpu = new CPU("Intel Core i7", 35000, 8, 3.8);
RAM ram = new RAM("Kingston Fury", 8000, 16, "DDR4");
GPU gpu = new GPU("NVIDIA RTX 4070", 60000, 12, "PCIe 4.0");
HDD hdd = new HDD("Seagate Barracuda", 5000, 2000, "SATA III");

Console.WriteLine("Проверка ToString:");
Console.WriteLine($"  {cpu.ToString()}\n");

Console.WriteLine("Проверка операторов сравнения:");
CPU cpuCopy = new CPU("Intel Core i7", 35000, 8, 3.8);
CPU cpuOther = new CPU("AMD Ryzen 7", 30000, 8, 3.6);
Console.WriteLine($"  cpu == cpuCopy: {cpu == cpuCopy}");
Console.WriteLine($"  cpu != cpuOther: {cpu != cpuOther}\n");

Console.WriteLine("Установка компонентов:");
Computer pc = new Computer();
pc.AddComponent(cpu);
pc.AddComponent(ram);
pc.AddComponent(gpu);
pc.AddComponent(hdd);

Console.WriteLine("\nИнформация о сборке:");
pc.ShowAllComponents();

Console.WriteLine($"\nИтого: {pc.GetTotalPrice()} руб.\n");

Console.WriteLine("Удаление видеокарты:");
pc.RemoveComponent(gpu);

Console.WriteLine($"\nНовая стоимость: {pc.GetTotalPrice()} руб.");
```
