Создать базовый класс Customer в C#, который будет представлять информацию о
клиентах или покупателях. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

// баз класс
public class Customer
{
    public int CustomerId { get; set; }
    public string Name { get; set; }
    public string Email { get; set; }

    public Customer(int customerId, string name, string email)
    {
        CustomerId = customerId;
        Name = name;
        Email = email;
    }

    public virtual string GetFullName() => Name;
    public virtual void UpdateEmail(string newEmail) => Email = newEmail;
    public virtual string ViewProfile() => $"ID: {CustomerId}, Имя: {GetFullName()}, Email: {Email}";
}

// вип
public class VipCustomer : Customer
{
    public int LoyaltyPoints { get; set; }

    public VipCustomer(int customerId, string name, string email, int points) : base(customerId, name, email)
    {
        LoyaltyPoints = points;
    }

    public override string ViewProfile() => base.ViewProfile() + $", Баллы лояльности: {LoyaltyPoints}, Статус: VIP";
}

// обыч клиент 
public class RegularCustomer : Customer
{
    public DateTime RegistrationDate { get; set; }
    public DateTime? LastEmailUpdate { get; private set; }

    public RegularCustomer(int customerId, string name, string email, DateTime regDate) : base(customerId, name, email)
    {
        RegistrationDate = regDate;
    }

    public override void UpdateEmail(string newEmail)
    {
        base.UpdateEmail(newEmail);
        LastEmailUpdate = DateTime.Now;
    }

    public override string ViewProfile()
    {
        string info = base.ViewProfile() + $", Дата регистрации: {RegistrationDate:dd.MM.yyyy}";
        if (LastEmailUpdate.HasValue)
            info += $", Последнее обновление email: {LastEmailUpdate:dd.MM.yyyy HH:mm:ss}";
        return info;
    }
}

// груп клиентт
public class GroupCustomer : Customer
{
    public string GroupName { get; set; }

    public GroupCustomer(int customerId, string name, string email, string groupName) : base(customerId, name, email)
    {
        GroupName = groupName;
    }

    public override string GetFullName() => $"Группа \"{GroupName}\" (контакт: {Name})";
}

// прог
var customers = new Customer[]
{
    new VipCustomer(1, "Иван Петров", "ivan@mail.ru", 1500),
    new RegularCustomer(2, "Мария Сидорова", "maria@mail.ru", new DateTime(2023, 5, 12)),
    new GroupCustomer(3, "ООО Ромашка", "contact@romashka.ru", "Оптовая закупка")
};

// Обновляем email Марии ДО вывода профилей
var regular = (RegularCustomer)customers[1];
regular.UpdateEmail("maria.new@mail.ru");

// вывод проф 
foreach (var c in customers)
{
    Console.WriteLine(c.ViewProfile());
    Console.WriteLine($"Полное имя: {c.GetFullName()}");
}
```
