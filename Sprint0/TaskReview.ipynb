```
using System;

public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветёт.");
    }

    public void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"Растение {Name} выросло. Новая высота: {Height} см.");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, float height, string color, int petalCount)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Wilt()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать.");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, float height, int age, string leafType)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowYears(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} теперь в возрасте {Age} лет.");
    }
}

Flower flower = new Flower("Тюльпан", 25, "Жёлтый", 6);
Tree tree = new Tree("Клён", 180, 15, "Широколиственные");

Console.WriteLine("- Цветок -");
Console.WriteLine($"Имя: {flower.Name}");
Console.WriteLine($"Тип: {flower.Type}");
Console.WriteLine($"Высота: {flower.Height} см");
Console.WriteLine($"Цвет: {flower.Color}");
Console.WriteLine($"Количество лепестков: {flower.PetalCount}");
flower.Bloom();
flower.Grow(5);
flower.Wilt();

Console.WriteLine("\n- Дерево -");
Console.WriteLine($"Имя: {tree.Name}");
Console.WriteLine($"Тип: {tree.Type}");
Console.WriteLine($"Высота: {tree.Height} см");
Console.WriteLine($"Возраст: {tree.Age} лет");
Console.WriteLine($"Тип листьев: {tree.LeafType}");
tree.Bloom();
tree.Grow(20);
tree.GrowYears(3);
```
